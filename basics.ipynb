##Basics of numpy and arrays


```python
import numpy as np  #importing numpy library
array=np.array([1,2,3,4,5])
print(array)
##lets perfrom operation on the array
array2=array*2
print("After multiplying by Two",array2)
print("After adding 5",array+5)
print("After subtracting 3",array-3)
print("After dividing by 2",array/2)
print(type(array))
```
