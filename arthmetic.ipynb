#Scalar arthmetics in numpy  (singel value)

```python
import numpy as np
np.array([1, 2, 3])
print("Addition of 2 in each element of array is: ", np.array([1, 2, 3]) + 2)
print("Multiplication of 2 in each element of array is: ", np.array([1, 2, 3]) * 2 )
print("Subtraction of 2 in each element of array is: ", np.array([1, 2, 3]) - 2)
print("Division of 2 in each element of array is: ", np.array([1, 2, 3]) / 2)       
print("Square of each element of array is: ", np.array([1, 2, 3]) ** 5)
```

#vectorize Math functions in numpy

```python
import numpy as np
array=np.array([1, 2, 3, 4, 5])
print(np.sqrt(array))

array1=np.array([1.9, 2.5, 3.7])
print("Original array:", array1)
print("ROUNDED VALUE:",np.round(array1))   #this will round the values to the nearest integer
print("CEIL VALUE:",np.ceil(array1))  #this will round up the values
print("Value of Pi:", np.pi)  

#LETS CALULATE THE REDIUS
radii = np.array([1, 2, 3, 4, 5])
area=np.pi *radii **2
print("Area of circles:", area)
```

#element wise arthmetic

```python
import numpy as np
array1=np.array([1,2,3])
array2=np.array([4,5,6])
print("Array 1 is: ", array1)
print("Array 2 is: ", array2)
addition = np.add(array1, array2)
print("Addition of two arrays is: ", addition)
subtraction = np.subtract(array1, array2)
print("Subtraction of two arrays is: ", subtraction)
multiplication = np.multiply(array1, array2)
print("Multiplication of two arrays is: ", multiplication)
division = np.divide(array1, array2)    
print("Division of two arrays is: ", division)
power = np.power(array1, array2)
print("Power of two arrays is: ", power)
```

#comparsion operator

```python
import numpy as np

array = np.array([55, 66, 82, 90, 69, 72])

print("Original array is:", array)
print("array == 69:", array == 69)   #this will give a boolean array where the condition is true or false
print("array > 70:", array > 70)     

with open("grades.txt", "w") as f:
    for marks in array:
        if marks == 90:
            grade = "A+ GRADE"
        elif marks == 82:
            grade = "A GRADE"
        elif marks > 70:
            grade = "B GRADE"
        elif marks == 69:
            grade = "D+ GRADE"
        elif marks < 60:
            grade = "F GRADE"
        else:
            grade = "PASS"

        f.write(f"Marks: {marks}::Grade: {grade}\n")

print("Grades have been written to grades.txt file.")

with open("grades.txt", "r") as f:
    content = f.read()
    print("Content of grades.txt file:")
    print(content)

```
