```python
import numpy as np
Ages=np.array([[18,24,25,23,65,12,13,15],
              [20,99,88,23,55,9,18,14]])
teenagers=Ages[Ages<18]
print("Teenagers:",teenagers)
adults=Ages[Ages>=18]
print("Adults:",adults)
aged_18_65=Ages[(Ages>=18) & (Ages<=65)]
print("Aged between 18 and 65:",aged_18_65)
#where function:to preserve the original shape of the array we can use where function
adults=np.where(Ages>=18,Ages,0) #now give 2d array with 0 where condition is not satisfied
print("Adults:\n",adults)
```
