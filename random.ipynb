#random numbers through numpy

```python
import numpy as np
random_numbers = np.random.default_rng()
print(random_numbers.integers(1, 100))
```

```python
import numpy as np
random_numbers = np.random.default_rng()
print("Number Guessing Game:")
print("Guess a number between 1 and 100.")
guess = int(input("Enter your guess: "))
target_number = random_numbers.integers(1, 100)
if guess == target_number:
    print("Congratulations! You guessed the correct number:", target_number)
else:
    print("Sorry, the correct number was:", target_number)
```
